```python
#creating a tuple
t=(10,20,30,40,5)

#Indexing
print("Element at index:",t[0])

#Negative Indexing
print("Element at list index:",t[-1])

#Slicing
print("Element from index 1:to3:",t[1:4]) # Corrected slicing for elements from index 1 to 3

#Iteration
print("Iterating elements:")
for i in t:
  print(i)
```
